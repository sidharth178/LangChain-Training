Text splitting is one of the most important steps in any LangChain workflow that uses documents, PDFs, or long text.

Why do we split text?
- Large documents may exceed the model's context window.
- Retrieval systems work better when each chunk is small and focused.
- Smaller chunks improve similarity search and reduce noise.

A text splitter decides how a long document is broken into manageable pieces. The most common choices are:
- `CharacterTextSplitter`: cuts by characters.
- `RecursiveCharacterTextSplitter`: tries to keep logical structure like sentences and paragraphs.

### 1. Text Splitter

Key settings to remember:
- `chunk_size`: maximum length of each chunk.
- `chunk_overlap`: how much one chunk overlaps the next.
- `separator`: the boundary used for splitting.

This is especially useful in retrieval-augmented generation (RAG), where we index document chunks and ask the model questions about them.

### 2. When to use text splitters

Use a text splitter when:
- you are working with long documents or PDFs,
- you want to build a RAG pipeline,
- you need to keep each chunk semantically relevant,
- you want to reduce token usage while preserving information.

Common mistakes:
- setting `chunk_overlap` too high, which creates repeated content,
- using a tiny `chunk_size`, which breaks context too aggressively,
- splitting by raw characters without thinking about natural text boundaries.

In practice, the best splitter depends on the kind of text you are processing:
- code -> better with a code-aware splitter,
- markdown -> recursive splitting works well,
- plain prose -> paragraph/sentence-aware splitting is often ideal.

````python
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader('D:\\LangChain\\Data\\RESUME 25MAS10038.pdf')

docs = loader.load()

splitter = CharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=0,
    separator=''
)

result = splitter.split_documents(docs)

print(result[1].page_content)
````

### 2. Recursive Character TextSplitter

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """
Space exploration has led to incredible scientific discoveries. From landing on the Moon to exploring Mars, humanity continues to push the boundaries of what’s possible beyond our planet.

These missions have not only expanded our knowledge of the universe but have also contributed to advancements in technology here on Earth. Satellite communications, GPS, and even certain medical imaging techniques trace their roots back to innovations driven by space programs.
"""

# Initialize the splitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=0,
)

# Perform the split
chunks = splitter.split_text(text)

print(len(chunks))
print(chunks)
````

### 3.2. Recursive Character TextSplitter - MARKDOWN

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

text = """
# Project Name: Smart Student Tracker

A simple Python-based project to manage and track student data, including their grades, age, and academic status.


## Features

- Add new students with relevant info
- View student details
- Check if a student is passing
- Easily extendable class-based design


## 🛠 Tech Stack

- Python 3.10+
- No external dependencies


## Getting Started

1. Clone the repo  
   ```bash
   git clone https://github.com/your-username/student-tracker.git

"""

# Initialize the splitter
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.MARKDOWN,
    chunk_size=200,
    chunk_overlap=0,
)

# Perform the split
chunks = splitter.split_text(text)

print(len(chunks))
print(chunks[0])
````

### 3.3. Recursive CharacterTextSplitter - Python

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

text = """
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade  # Grade is a float (like 8.5 or 9.2)

    def get_details(self):
        return self.name"

    def is_passing(self):
        return self.grade >= 6.0


# Example usage
student1 = Student("Aarav", 20, 8.2)
print(student1.get_details())

if student1.is_passing():
    print("The student is passing.")
else:
    print("The student is not passing.")

"""

# Initialize the splitter
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=300,
    chunk_overlap=0,
)

# Perform the split
chunks = splitter.split_text(text)

print(len(chunks))
print(chunks[1])
````
