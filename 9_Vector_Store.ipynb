### Vector Stores in LangChain

#### What is a vector store?
A vector store is a storage layer used to keep text chunks as embeddings, so the system can retrieve relevant information by semantic similarity rather than by exact keyword matching.

This is useful for:
- semantic search
- Retrieval-Augmented Generation (RAG)
- document similarity and clustering
- question-answering over a knowledge base

#### Vector store vs vector database
This is a common point of confusion:

- Vector store: the high-level concept or interface used by LangChain applications to store and search embeddings.
- Vector database: the actual backend technology that indexes vectors and supports fast nearest-neighbor search.

In simple terms:
- the vector store is the app-facing abstraction
- the vector database is the engine underneath

Example:
- LangChain may use a `FAISS` vector store in your app
- `FAISS` is also a vector database/indexing system
- Chroma, Pinecone, Qdrant, and Weaviate are also vector databases that can be wrapped by LangChain vector stores

So, in practice, a vector store in LangChain often sits on top of a vector database or indexing library.


#### How it works
Each document is converted to an embedding vector using a model such as OpenAI embeddings or Hugging Face embeddings. These vectors are stored in the vector database.

When a user asks a question:
- the question is turned into an embedding
- the vector store compares it with stored vectors
- the closest matches are returned as relevant context

This is often called semantic retrieval.

#### Common vector store operations
- `add_documents(...)`: store new chunks
- `similarity_search(...)`: find nearest matches by semantic similarity
- `similarity_search_with_score(...)`: return relevance score
- `as_retriever(...)`: use it directly in a LangChain retriever pipeline

#### Examples of vector stores
- FAISS (fast local in-memory/vector index)
- Chroma (easy local or persistent vector DB)
- Pinecone (managed cloud vector DB)
- Weaviate, Qdrant, pgvector

#### When to use a vector store
Use a vector store when you need retrieval based on meaning rather than exact string matching. It is especially helpful for:
- PDFs, docs, notes, FAQs, internal knowledge bases
- chatbot memory over documents
- RAG systems that ground answers in context

#### Common mistakes to avoid
- Using a vector store without embeddings: it cannot compare meaning
- Forgetting to split large documents into chunks
- Not storing metadata like source, title, or timestamp when needed
- Overloading the context window with too many retrieved chunks
- Mixing large documents with very small chunk sizes without tuning

```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_core.documents import Document

# Create LangChain documents for IPL players

doc1 = Document(
        page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and fitness, he has led the Royal Challengers Bangalore in multiple seasons.",
        metadata={"team": "Royal Challengers Bangalore"}
    )
doc2 = Document(
        page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm demeanor and ability to play big innings under pressure.",
        metadata={"team": "Mumbai Indians"}
    )
doc3 = Document(
        page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicketkeeping, and leadership are legendary.",
        metadata={"team": "Chennai Super Kings"}
    )
doc4 = Document(
        page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his yorkers and death-over expertise.",
        metadata={"team": "Mumbai Indians"}
    )
doc5 = Document(
        page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and match-winning performances make him a key player.",
        metadata={"team": "Chennai Super Kings"}
    )

docs = [doc1, doc2, doc3, doc4, doc5]

vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory='my_chroma_db',
    collection_name='sample'
)
```

```python
# add documents
vector_store.add_documents(docs)
```

```python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python
# search documents
vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2
)
```

```python
# search with similarity score
vector_store.similarity_search_with_score(
    query='Who among these are a bowler?',
    k=2
)
```

```python
# meta-data filtering
vector_store.similarity_search_with_score(
    query="",
    filter={"team": "Chennai Super Kings"}
)
```

```python
# update documents
updated_doc1 = Document(
    page_content="Virat Kohli, the former captain of Royal Challengers Bangalore (RCB), is renowned for his aggressive leadership and consistent batting performances. He holds the record for the most runs in IPL history, including multiple centuries in a single season. Despite RCB not winning an IPL title under his captaincy, Kohli's passion and fitness set a benchmark for the league. His ability to chase targets and anchor innings has made him one of the most dependable players in T20 cricket.",
    metadata={"team": "Royal Challengers Bangalore"}
)

vector_store.update_document(document_id='09a39dc6-3ba6-4ea7-927e-fdda591da5e4', document=updated_doc1)
```

```python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python
# delete document
vector_store.delete(ids=['09a39dc6-3ba6-4ea7-927e-fdda591da5e4'])
```
